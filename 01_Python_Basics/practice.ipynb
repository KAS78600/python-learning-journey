```python
import sys

def main():
    data = list(map(int, sys.stdin.buffer.read().split()))
    n, m = data[0], data[1]

    received_from = [set() for _ in range(n + 1)]
    gave_gift = [False] * (n + 1)

    index = 2
    for _ in range(m):
        giver, receiver = data[index], data[index + 1]
        index += 2

        gave_gift[giver] = True
        received_from[receiver].add(giver)

    for member in range(1, n + 1):
        if not gave_gift[member] and len(received_from[member]) == n - 1:
            print(member)
            return

    print(-1)

if __name__ == "__main__":
    main()
```
