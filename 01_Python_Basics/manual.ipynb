## Objective 9: Calculate Basic Statistics

Create a list of numbers and calculate its count, sum, mean, median, and mode.

```python
from collections import Counter
from statistics import mean, median

values = [4, 7, 7, 10, 12, 14]
value_counts = Counter(values)
mode_value = value_counts.most_common(1)[0][0]

print("Count:", len(values))
print("Sum:", sum(values))
print("Mean:", mean(values))
print("Median:", median(values))
print("Mode:", mode_value)
```

## Objective 10: Calculate Range, Variance, and Standard Deviation

Calculate and print the range, sample variance, and sample standard deviation of a list of numbers.

```python
from statistics import stdev, variance

values = [4, 7, 7, 10, 12, 14]
value_range = max(values) - min(values)

print("Range:", value_range)
print("Sample variance:", variance(values))
print("Sample standard deviation:", stdev(values))
```

## Objective 12: Linear Regression with scikit-learn

Fit a linear regression model to a small set of data points and print the regression coefficient and intercept.

```python
import numpy as np
from sklearn.linear_model import LinearRegression

x_values = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y_values = np.array([3, 5, 7, 9, 11])

linear_model = LinearRegression()
linear_model.fit(x_values, y_values)

print("Coefficient:", linear_model.coef_[0])
print("Intercept:", linear_model.intercept_)
```

## Objective 13: Linear Regression Summary with statsmodels

Fit an ordinary least squares regression model with statsmodels and print the full regression summary table.

```python
import pandas as pd
import statsmodels.api as sm

regression_data = pd.DataFrame({
    "hours_studied": [1, 2, 3, 4, 5, 6, 7, 8],
    "exam_score": [52, 57, 65, 68, 75, 78, 86, 90]
})

x_values = sm.add_constant(regression_data["hours_studied"])
y_values = regression_data["exam_score"]
regression_model = sm.OLS(y_values, x_values).fit()
print(regression_model.summary())
```

## Objective 14: Print Regression Coefficients and Model Statistics

Use statsmodels to fit a linear regression model and print its regression coefficients, p-values, and R-squared value.

```python
import pandas as pd
import statsmodels.api as sm

regression_data = pd.DataFrame({
    "hours_studied": [1, 2, 3, 4, 5, 6, 7, 8],
    "exam_score": [52, 57, 65, 68, 75, 78, 86, 90]
})

x_values = sm.add_constant(regression_data["hours_studied"])
y_values = regression_data["exam_score"]
regression_model = sm.OLS(y_values, x_values).fit()

print("Regression coefficients:")
print(regression_model.params)
print("P-values:")
print(regression_model.pvalues)
print("R-squared:", regression_model.rsquared)
```

## Objective 15: Plot a Styled Line with Markers

Create a simple line graph with markers and customize the line style.

## Objective 16: Add Labels, Title, and Grid to a Line Graph

Plot a line graph with x-axis and y-axis labels, a title, and grid lines.

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
temperatures = [12, 14, 18, 22, 26]

fig, ax = plt.subplots()
ax.plot(months, temperatures, marker="o")
ax.set_xlabel("Month")
ax.set_ylabel("Temperature (C)")
ax.set_title("Monthly Temperatures")
ax.grid(True)
plt.show()
plt.close(fig)
```

## Objective 17: Create Two Subplots

Create one figure with two subplots and plot different data in each subplot.

```python
import matplotlib.pyplot as plt

x_values = [1, 2, 3, 4, 5]
first_data = [1, 4, 9, 16, 25]
second_data = [1, 8, 27, 64, 125]

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(x_values, first_data, marker="o", color="royalblue")
axes[0].set_title("Squares")
axes[1].plot(x_values, second_data, marker="s", color="darkorange")
axes[1].set_title("Cubes")
fig.tight_layout()
plt.show()
plt.close(fig)
```

## Objective 18: Create a Bar Chart with Value Labels

Create a Matplotlib bar chart and label each bar with its corresponding value.

```python
import matplotlib.pyplot as plt

fruits = ["Apples", "Bananas", "Cherries", "Dates"]
quantities = [12, 8, 15, 6]

fig, ax = plt.subplots()
bars = ax.bar(fruits, quantities, color=["seagreen", "gold", "tomato", "slateblue"])
ax.bar_label(bars, padding=3)
ax.set_ylabel("Quantity")
ax.set_title("Fruit Quantities")
plt.show()
plt.close(fig)
```

## Objective 19: Generate Random Data and Plot a Histogram

Generate reproducible random data and display its distribution in a Matplotlib histogram.

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
random_values = np.random.normal(loc=50, scale=10, size=500)

fig, ax = plt.subplots()
ax.hist(random_values, bins=20, color="cornflowerblue", edgecolor="black")
ax.set_xlabel("Value")
ax.set_ylabel("Frequency")
ax.set_title("Histogram of Random Data")
plt.show()
plt.close(fig)
```

## Objective 20: Create a Pie Chart with Labels and Percentages

Create a Matplotlib pie chart with a label and percentage shown for each slice.

```python
import matplotlib.pyplot as plt

categories = ["Python", "Java", "C++", "JavaScript"]
values = [40, 25, 20, 15]

fig, ax = plt.subplots()
ax.pie(values, labels=categories, autopct="%1.1f%%", startangle=90)
ax.set_title("Programming Language Preferences")
ax.axis("equal")
plt.show()
plt.close(fig)
```
